**Fill with methods**

- whenenver we are using methods no need to provide values

- pad
- bfill
- ffill
- backfill

```python
import numpy as np
import pandas as pd
Data={'Names':['Ramesh','Suresh','Sathish',np.nan],
     'Age':[np.nan,22,25,27],
     'City':['Pune','Pune',np.nan,'Hyd']}

df=pd.DataFrame(Data)
df
```

```python
import warnings
warnings.filterwarnings('ignore')
df.fillna(method='pad')
```

```python
import warnings
warnings.filterwarnings('ignore')
df.fillna(method='pad',axis=1)
```

```python
import warnings
warnings.filterwarnings('ignore')
df.fillna(method='bfill')
```

```python
import warnings
warnings.filterwarnings('ignore')
df.fillna(method='bfill',axis=1)
```

```python
import warnings
warnings.filterwarnings('ignore')
df.fillna(method='ffill')
```

```python
import warnings
warnings.filterwarnings('ignore')
df.fillna(method='ffill',axis=1)
```

```python
import warnings
warnings.filterwarnings('ignore')
df.fillna(method='backfill')
```

```python
import warnings
warnings.filterwarnings('ignore')
df.fillna(method='backfill',axis=1)
```

**Mean-Median-Mode**
- instead of filling with random values
- we can use mean median 

```python
age_mean=df['Age'].mean()
df['Age'].fillna(age_mean)
```

```python
age_median=df['Age'].median()
df['Age'].fillna(age_median)
```

**KNN Immputer**

- K-Nearest Neighbours
- Where k=hyper parameters which means the user can change
- Idea: Instead of taking all the samples average
    - First fix the neighbours number i.e. k
    - Then find the neighbours using distance metric
    - Then take the average of those samples to fill the missing value

```python
import sklearn
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import LabelEncoder
from sklearn.impute import KNNImputer
```

```python
import numpy as np
import pandas as pd
from sklearn.impute import KNNImputer
x=[[1,2,np.nan],[3,np.nan,3],[np.nan,60,5],[8,8,7]]
pd.DataFrame(x)
```

```python
import numpy as np
import pandas as pd
from sklearn.impute import KNNImputer
x=[[1,2,np.nan],[3,np.nan,3],
   [np.nan,60,5],[8,8,7]]
imputer=KNNImputer(n_neighbors=2,
                  weights='uniform')
imputer.fit_transform(x)
```

```python
import numpy as np
import pandas as pd
from sklearn.impute import KNNImputer
x=[[1,2,np.nan],[3,np.nan,3],
   [np.nan,60,5],[8,8,7]]
imputer=KNNImputer(n_neighbors=2,
                  weights='distance')
imputer.fit_transform(x)
```
