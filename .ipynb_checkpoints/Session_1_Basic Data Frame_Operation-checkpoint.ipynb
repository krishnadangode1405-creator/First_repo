**Step-8: Drop the column**

- label
- axis
- index
- columns
- inplace

```python
df.drop('age')   # error
# here python assume age is a label
# python ask qn: is it index label or column label
# axis=1 means for columns
# axis=0 means for rows
# by default axis=0
```

```python
df.drop('age',axis=0) # error
```

```python
df.drop('age',
        axis=1)  # corrected
```

```python
# we are able to see old columns
# because we did not apply inplace=True
```

```python
# whenever we use columns
# no need of axis

df.drop(columns=['age'],
        inplace=True)
```

```python
# label
df.drop(2,
       axis=0)
```

```python
df1.loc[3:7,['square_value','cube_value']]
```

```python
cols=['square_value','cube_value']
df1.loc[3:7,cols]
```

```python
df1.iloc[[2,3,4]]  
# selectiong 3 rows ,all the columns
# dataframe
```

```python
df1.iloc[[2]]  # 2D dataframe
```

```python
df1.iloc[2]  # 1D series
```

```python
df1.iloc[[2,3,4],[0,2]]
# 2,3,4 row
# 0 and 2nd index (1,3)
```

```python
df1.iloc[2,:]
# 2nd row ,all column
```
