```python
# step-1
for i in cat[1:]:
    keys=visa_df[i].value_counts().keys()
    value=visa_df[i].value_counts().values
    df=pd.DataFrame(zip(keys,value),
                   columns=['Label','No of applicants'])
    
    # step-2
    folder_name='frequncy_table'
    os.makedirs(folder_name,exist_ok=True)
    
    # step-3
    cwd=os.getcwd()
    new_dir=os.path.join(cwd,folder_name)
    file_name=f'{i}.csv'
    path=os.path.join(new_dir,file_name)
    
    # step-4
    df.to_csv(path,index=False)
```

**Bar Chart Automate**

```python
for i in cat[1:]:
    keys=visa_df[i].value_counts().keys()
    value=visa_df[i].value_counts().values
    plt.figure(figsize=(10,4))
    
    plt.bar(keys,value)
    folder_name='Bar_Chart'
    os.makedirs(folder_name,exist_ok=True)
    
    cwd=os.getcwd()
    new_dir=os.path.join(cwd,folder_name)
    file_name=f'{i}.jpg'
    path=os.path.join(new_dir,file_name)
    
    plt.xlabel('labels')
    plt.ylabel('count')
    plt.title(f'{i}_barchart')
    plt.savefig(path)
    plt.show()
```

**Pie Chart**

```python
for i in cat[1:]:
    keys=visa_df[i].value_counts().keys()
    values=visa_df[i].value_counts().values
    plt.pie(values,
            explode=[0.1 for i in range(len(keys))],
            labels=keys,
            autopct='%0.2f%%',
            startangle=45,
            radius=2)
    folder_name='Pie_Chart'
    os.makedirs(folder_name,exist_ok=True)
    
    cwd=os.getcwd()
    new_dir=os.path.join(cwd,folder_name)
    file_name=f'{i}.jpg'
    path=os.path.join(new_dir,file_name)
    
    plt.title(f'{i}_piechart')
    plt.savefig(path)
    plt.show()
```
